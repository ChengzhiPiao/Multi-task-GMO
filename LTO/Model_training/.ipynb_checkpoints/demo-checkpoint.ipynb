```python
import torch
val_dataset = torch.load('./val_dataset_OVERLAP_git.pt')
```

```python
len(val_dataset['mcs'])
```

```python
import csv
f=open("ground_truth_OVERLAP_git.txt","w")
num=len(val_dataset['mcs'])
f.write(str(num)+"\n")
for i in range(0,num):
    f.write(str(val_dataset['mcs'][i])+"\n")
f.close()
f = open("val_data_OVERLAP_git.txt", "w")
f.write(str(num)+"\n")
train_num=125432
test_num=36262
val_num=36149
for i in range(0,num):
    f.write(str(val_dataset['soft_matrix'][i].size(0))+" "+str(val_dataset['soft_matrix'][i].size(1))+"\n")
    for j in range(0,val_dataset['soft_matrix'][i].size(0)):
        Mapping_matrix=val_dataset['soft_matrix'][i][j].detach().numpy()
        for k in range(0,val_dataset['soft_matrix'][i].size(1)):
            f.write(str(Mapping_matrix[k])+" ")
        f.write("\n")
    for j in range(0,val_dataset['soft_matrix'][i].size(0)):
        mapping_matrix=val_dataset['mapping'][i][j].detach().numpy()
        for k in range(0,val_dataset['soft_matrix'][i].size(1)):
            f.write(str(int(mapping_matrix[k]))+" ")
        f.write("\n")
    graphAid = val_dataset["graphAid"][i]
    if graphAid<train_num:
        with open(f'./json_data/git/train/{graphAid}.csv', mode='r') as file:
            g1 = csv.reader(file)
            g1=list(g1)
    elif graphAid<train_num+test_num:
        with open(f'./json_data/git/test/{graphAid}.csv', mode='r') as file:
            g1 = csv.reader(file)
            g1=list(g1)
    else:
        with open(f'./json_data/git/val/{graphAid}.csv', mode='r') as file:
            g1 = csv.reader(file)
            g1=list(g1)
    graphBid = val_dataset["graphBid"][i]
    if graphBid<train_num:
        with open(f'./json_data/git/train/{graphBid}.csv', mode='r') as file:
            g2 = csv.reader(file)
            g2=list(g2)
    elif graphBid<train_num+test_num:
        with open(f'./json_data/git/test/{graphBid}.csv', mode='r') as file:
            g2 = csv.reader(file)
            g2=list(g2)
    else:
        with open(f'./json_data/git/val/{graphBid}.csv', mode='r') as file:
            g2 = csv.reader(file)
            g2=list(g2)
    r=len(g1)
    c=len(g1[0])
    f.write(str(r)+" "+str(c)+"\n")
    value=dict()
    cnt=0
    for j in range(r):
        for k in range(c):
            if g1[j][k]==None:
                g1[j][k]="#$%^&^%"
            if g1[j][k] not in value:
                cnt+=1
                value[g1[j][k]]=cnt
            f.write(str(value[g1[j][k]])+" ")
        f.write("\n")
    r=len(g2)
    c=len(g2[0])
    f.write(str(r)+" "+str(c)+"\n")
    for j in range(r):
        for k in range(c):
            if g2[j][k]==None:
                g2[j][k]="#$%^&^%"
            if g2[j][k] not in value:
                cnt+=1
                value[g2[j][k]]=cnt
            f.write(str(value[g2[j][k]])+" ")
        f.write("\n")
f.close()
```
